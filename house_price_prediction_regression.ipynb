**Decision Insight**:
- **LotArea** , **3SsnPorch** , **LowQualFinSF** , **BsmtFinSF2** , **ScreenPorch** , **EnclosedPorch** , **OpenPorchSF** are having large outliers. Means Most of the House are having same Quality of Features. Very Few Houses having Very High Features in the House.

- We have to apply PowerTransformation for reducing the variance Robust Scaler Technique. Beacause it is not effectly by the outliers

```python
from sklearn.preprocessing import PowerTransformer

# remove the target variable in skewness df

skewed_cols_df = numerical_analysis_report_df.drop(
  numerical_analysis_report_df[numerical_analysis_report_df['Columns']=='SalePrice'].index,
  axis=0
)

skewed_cols_df = skewed_cols_df.drop(
  skewed_cols_df[skewed_cols_df['Skewness']<1].index,
  axis=0,
)

# Applying power transformation to all the skewed columns
for col in skewed_cols_df['Columns'].tolist():
  power_transformer = PowerTransformer(method='yeo-johnson')
  df[f'power_{col}'] = (power_transformer.fit_transform(df[[col]]).ravel())
```

```python
# Finding the Difference between Skewed Columns with Power_transformed columns

fig, ax = plt.subplots(6, 4, figsize=(20, 20))

ax = ax.flatten()

idx = 0

for col in skewed_cols_df['Columns'].tolist():

    # Original distribution
    sns.histplot(df[col].dropna(), ax=ax[idx], kde=True)
    ax[idx].set_title(f'Original: {col}')
    idx += 1

    # Power transformed distribution
    sns.histplot(df[f'power_{col}'].dropna(), ax=ax[idx], kde=True)
    ax[idx].set_title(f'Power Transformed: {col}')
    idx += 1
    if idx >= 24:
        break

plt.tight_layout()
plt.show()
```

**Handling Multicollinearity Among Categorical Columns with Cramers'v Value and Chi-Square Test**

```python
cramers_matrix = pd.DataFrame(0.0,columns=categorical_columns,index=categorical_columns)


# calculate Cramers Value
def calculate_cramers_value(chi2,table):
  n = table.to_numpy().sum()
  r,c = table.shape
  cramers_value = np.sqrt(chi2/(n*min(r-1,c-1)))
  return cramers_value


for col1 in categorical_columns:
  for col2 in categorical_columns:
    if col1 == col2:
      cramers_matrix.loc[col1,col2] = 1.0
      continue
    else:
      pair = df[[col1,col2]].dropna()
      table = pd.crosstab(pair[col1],pair[col2])
      chi2,p_val,_,_ = chi2_contingency(table)
      cramers_value = calculate_cramers_value(chi2,table)
      cramers_matrix.loc[col1,col2] = cramers_value

```

```python
def Identify_multicollinearity_columns(corr_df,threshold=0.8) -> pd.DataFrame:
  correlated_pairs = []
  
  for col1 in range(corr_df.shape[1]):
    for col2 in range(col1 + 1,corr_df.shape[1]):
      
      cramers_val = corr_df.iloc[col1,col2]
      
      if abs(cramers_val) >= threshold:
        correlated_pairs.append({
          'Col1' : corr_df.columns[col1],
          'Col2' : corr_df.columns[col2],
          'Correlation' : cramers_val
        })
  
  cat_correlated_cols_df = pd.DataFrame(correlated_pairs)
  cat_correlated_cols_df = cat_correlated_cols_df.sort_values('Correlation',ascending=False)
  return cat_correlated_cols_df
```

```python
# Calculate Mannwhitneyu Test Effect Size
def calculate_mannwhitneyu_effectsize(u_stat,groups):
  group1 ,group2 = groups

  N = len(group1) + len(group2)
  
  mean_u = (len(group1) * len(group2)) / 2
  
  std_u = np.sqrt((len(group1) * len(group2)) * (len(group1) + len(group2) + 1)/12)
  
  z = (u_stat - mean_u) / std_u
  
  r_val = abs(z) / np.sqrt(N)
  
  return r_val



# calculate Kruskal's Test Effect size
def calculate_kruskal_effectsize(H_val,groups):
  
  N = sum(len(group) for group in groups)
  
  k = len(groups)
  
  epsilon_squared = (H_val - k + 1) / (N - k)
  
  return epsilon_squared

```

**`🎯 Model Building`**

**Model Evaluation custom Function**

```python
def model_evaluation(y_true,y_pred):
  r2 = r2_score(y_true,y_pred)
  mae = mean_absolute_error(y_true,y_pred)
  mse = mean_squared_error(y_true,y_pred)
  rmse = np.sqrt(mse)
  return r2,mae,mse,rmse
```

**Models Building Custom Function**

```python
model_comparison = []
```

```python
def model_building(models_dicts:dict):
  train_r2_scores = []
  test_r2_scores = []
  for model_name,model in models_dicts.items():
    # Training the Model
    model.fit(x_train_preprocessed,y_train)
    
    # Train Predictions
    y_train_predict = model.predict(x_train_preprocessed)
    
    # Test Prediction
    y_test_predict = model.predict(x_test_preprocessed)
    
    # Training performance
    train_r2,train_mae,train_mse,train_rmse = model_evaluation(y_train,y_train_predict)
    print(f'------ {model_name} -------')
    print(f'- Training Performance : ')
    print(f'R2-Squared : {train_r2}')
    print(f'Mean-Squared Error : {train_mse}')
    print(f'Mean-Absolute Error : {train_mae}')
    print(f'Root Mean-Squared Error : {train_rmse}')
    print()
    # Testing Performance
    test_r2,test_mae,test_mse,test_rmse = model_evaluation(y_test,y_test_predict)
    print(f'- Testing Performance : ')
    print(f'R2-Squared : {test_r2}')
    print(f'Mean-Squared Error : {test_mse}')
    print(f'Mean-Absolute Error : {test_mae}')
    print(f'Root Mean-Squared Error : {test_rmse}')
    print(f'='*40)
    
    
    model_comparison.append({
      'Model' : model_name,
      'train_r2' : train_r2,
      'train_mae' : train_mae,
      'train_mse' : train_mse,
      'train_rmse' : train_rmse,
      'test_r2' : test_r2,
      'test_mae' : test_mae,
      'test_mse' : test_mse,
      'test_rmse' : test_rmse
    })
    
    train_r2_scores.append(train_r2)
    test_r2_scores.append(test_r2)
    
  return train_r2_scores,test_r2_scores
  
```

**`🌍 Hyper-Parameter Tuning`**

```python
param_grids = {
  'LinearRegression' : {
    'fit_intercept' : [True,False],
    'positive' : [True,False]
  },
  'Ridge' : {
    'alpha' : np.logspace(-4,4,50)
  },
  'ElasticNet':{
    'alpha' : np.logspace(-5,1,20),
    'l1_ratio':[0.0,0.1,0.25,0.5,0.75,0.9,1.0]
  },
  'SVR':{
    'kernel':['rbf'],
    'C':[0.1,1,10,50,100,500],
    'epsilon':[0.01,0.05,0.1,0.2,0.5],
    'gamma':['scale','auto']
  }
}
```

```python
hyper_parameter_models ={
  'LinearRegression':LinearRegression(),
  'Ridge':Ridge(),
  'ElasticNet':ElasticNet(),
  'SVR':SVR()
}
```

```python
def hyper_parameter_tuning(model_list,param_grids):
  
  grid_models = {}
  
  for model_name,model in model_list.items():
    
    parameters_grid = param_grids[model_name]
    
    grid = GridSearchCV(
      estimator=model,
      param_grid=parameters_grid,
      cv=10,
      scoring='r2',
      n_jobs=-1
    )
    
    grid.fit(x_train_preprocessed,y_train)
    
    grid_models[model_name] = grid
    
    print(f'-- {model_name} -----')
    print(f'Best Parameters : ',grid.best_params_)
    print(f'Best CV Scores : ',grid.best_score_)
    print('\n','='*20)
    
```

```python
hyper_parameter_tuning(hyper_parameter_models,param_grids)
```

<h2 style='color:green;font-weight=bold;text-align:center'>Task 3: Buyer Recommendations</h2>


- Pick target neighborhoods first — location alone explains a large chunk of price variation.

- Within a neighborhood, rank homes by OverallQual first — it's the single biggest price lever.

- Compare price-per-square-foot of GrLivArea as your primary size metric.

- Treat basement size and garage capacity as separate value-adds, not bonuses.

- Check individual quality ratings (kitchen, exterior, basement, heating) — a good overall score can hide one weak spot.

- Central air conditioning carries a real, statistically-backed price premium — budget for it or negotiate its absence.

- Don't overpay for porches, screened porches, or pools — these showed weak correlation with price.

- Match bedroom/bathroom/room counts to actual needs — excess capacity adds cost without adding value to you.

- Use log-scale thinking: a given quality/size upgrade moves price by a roughly consistent percentage across price tiers.

- Sanity-check any listing's asking price against the model's predicted price using its OverallQual, GrLivArea, Neighborhood, and garage/basement size.
